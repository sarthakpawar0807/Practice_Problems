```python
# Program to print sum of all the digits of given number
num = int(input("Enter a number: "))
sum = 0
while num > 0:
    digit = num % 10
    sum = sum + digit
    num = num // 10
print("Sum of digits =", sum)
```

```python
# find the factorial of a given number
num = int(input("Enter a number: "))
Factorial = 1
for i in range(1 , num+1):
    Factorial = Factorial * i 
print("Factorial=" , Factorial)    
```

```python
# python program to check if given number prime or not
num = int(input("Enter a number: "))
if num % 2 == 0:
    for i in range(num-1 , num ):
        print("Number is not prime")
else:
    print("Number is prime")
    
```

```python
# python program to print prime numbers between 10 to 99
for num in range(10, 100):
    for i in range(2, num):
        if num % i == 0:
            break
    else:
        print(num)
```

```python
# ⁠Program to reverse a given number
num = int(input("Enter a number: "))
reverse = 0
while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10
print("Reverse =", reverse)
```

```python
# To determine whether a given year is a leap year or not.
Year = int(input("Enter the year"))
if Year % 4 == 0:
    print("Leap year")
else:
    print("Not a leap year")
```

```python
# In certain university, grades are assigned to the marks obtained in respective subjects.
marks = int(input("Enter marks in Math: "))
if marks >= 90:
    print("Grade = O")
elif marks >= 80:
    print("Grade = A+")
elif marks >= 70:
    print("Grade = A")
elif marks >= 60:
    print("Grade = B")
elif marks >= 50:
    print("Grade = C")
elif marks >= 40:
    print("Grade = P")
else:
    print("Grade = F")
```

```python
#⁠Accept Student  Name, Roll Number and Marks of the 3 subjects from the user. Calculate the percentageof 
#the marks and display Display the Subject with Highest and lowest marks
Student_name = input("Enter Student Name: ")
Roll_no = input("Enter Roll Number: ")
English = int(input("Enter marks of Subject 1: "))
Chemistry = int(input("Enter marks of Subject 2: "))
Maths = int(input("Enter marks of Subject 3: "))
total = English + Chemistry + Maths
percentage = total / 3
print("\nStudent Name:", Student_name)
print("Roll Number:", Roll_no)
print("Percentage:", percentage, "%")
highest = max(English , Chemistry , Maths)
lowest = min(English , Chemistry , Maths)
print("Highest Marks:", highest)
print("Lowest Marks:", lowest)
```
