```python
 """Accept 20 values from user and save it in list. Perform following operations on it
 a) count similar elements of list and print their index value   
b) count even and odd values of list 
c) count positive and negative values of list."""
Numbers = []
# Accept 20 values from user
for i in range(20):
    num = int(input("Enter a number: "))
    Numbers.append(num)
print("List:", Numbers)
# a) Count similar elements and print their index values
print("\nSimilar elements and their indexes:")
for num in Numbers:
    if Numbers.count(num) > 1:
        indexes = []
        for i in range(20):
            if Numbers[i] == num:
                indexes.append(i)
        print(num, "occurs", Numbers.count(num), "times at index", indexes)
# b) Count even and odd values
even = 0
odd = 0
for num in Numbers:
    if num % 2 == 0:
        even = even + 1
    else:
        odd = odd + 1
print("\nEven values:", even)
print("Odd values:", odd)
# c) Count positive and negative values
positive = 0
negative = 0
for num in Numbers:
    if num > 0:
        positive = positive + 1
    elif num < 0:
        negative = negative + 1
print("\nPositive values:", positive)
print("Negative values:", negative)
```

```python

"""Accept 10 values from user and save it in list. Perform following operations on it  
a) Sort list in ascending order using sorted() function and display sorted list
b) sort list in descending order using sort() function
c) display length of list"""
# Accept 10 values from user
Numbers = []
for i in range (10):
    num = int(input("Enter the number:"))
    Numbers.append(num)
print("Original List:", Numbers)    
# a) Sort in ascending order using sorted()
ascending = sorted(Numbers)
print("Ascending order:", ascending)
# b) Sort in descending order using sort()
Numbers.sort(reverse=True)
print("Descending order:", Numbers)
# c) Display length of list
print("Length of list:", len(Numbers))
```

```python
# Accept two lists from user and merge them using + in a single list
List1 = []
List2 = []
# Accept 5 values for List1
for i in range(5):
    num = int(input("Enter value for List1: "))
    List1.append(num)
# Accept 5 values for List2
for i in range(5):
    num = int(input("Enter value for List2: "))
    List2.append(num)
# Merge the two lists using +
List3 = List1 + List2
print("List 1:", List1)
print("List 2:", List2)
print("Merged List:", List3)
```

```python
# .Python program to remove duplicate elements from a list.
Numbers = [10, 20, 30, 20, 40, 10, 50, 30]
NewList = []
for num in Numbers:
    if num not in NewList:
        NewList.append(num)
print("Original List:", Numbers)
print("List after removing duplicates:", NewList)
```

```python
# Python program to find common elements in two lists.
List1 = [10, 20, 30, 40, 50]
List2 = [30, 40, 50, 60, 70]
Common = []
for num in List1:
    if num in List2:
        Common.append(num)
print("Common elements:", Common)
```
